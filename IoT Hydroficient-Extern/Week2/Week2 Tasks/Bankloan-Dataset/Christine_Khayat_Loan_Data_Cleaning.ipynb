### Inconsistent Values
I found negative values in the Experience column. Negative years of work experience are not logically valid, so I replaced all negative Experience values with 0. This avoids keeping impossible values while not assuming an exact number of years of experience.


```python
print("Education:", bank_loan["Education"].unique())
print("Personal Loan:", bank_loan["Personal Loan"].unique())
print("Securities Account:", bank_loan["Securities Account"].unique())
print("CD Account:", bank_loan["CD Account"].unique())
print("Online:", bank_loan["Online"].unique())
print("CreditCard:", bank_loan["CreditCard"].unique())
```

```python
bank_loan.describe()
```

## Identify Invalid Experience Values

I filtered the `Experience` column to find records containing negative values.

```python
bank_loan[bank_loan["Experience"] < 0]
```

## Correct Invalid Experience Values

I replaced all negative `Experience` values with 0.

```python
bank_loan.loc[bank_loan["Experience"] < 0, "Experience"] = 0
```

```python
bank_loan["Experience"].min()
#That confirms that there are no negative Experience values left.
```

```python
bank_loan.info()
```

```python
bank_loan.describe()
```

## Correct CCAvg Values

The `CCAvg` column was stored as an `object` data type and contained values such as `1/60` and `2/70`.

These values were intended to represent decimal numbers, so `/` was replaced with `.`.

```python
bank_loan["CCAvg"].head(20)
```

```python
bank_loan["CCAvg"] = bank_loan["CCAvg"].str.replace("/", ".", regex=False)
```

```python
bank_loan["CCAvg"] = pd.to_numeric(bank_loan["CCAvg"])
```
